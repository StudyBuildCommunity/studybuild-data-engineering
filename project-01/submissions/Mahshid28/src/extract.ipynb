```python
import os
import pandas as pd
from pathlib import Path
import logging
```

```python
BASE_DIR  = Path.cwd().parent
RAW_DIR   = BASE_DIR / "data" / "raw"
```

```python
logger = logging.getLogger(__name__)
```

```python
def load_single_file(raw_dir: Path, filename: str) -> Optional[pd.DataFrame]:
    """
    Load a single CSV file
    
    Args:
        filepath: Path to the CSV file
    
    Returns:
        DataFrame or None if file doesn't exist
    """
    filepath = raw_dir / filename
    if filepath.exists():
        try:
            df = pd.read_csv(filepath)
            logger.info(f"Loaded {filepath.name}: {len(df)} rows")
            return df
        except Exception as e:
            logger.error(f"Error loading {filepath.name}: {str(e)}")
            return None
    else:
        logger.warning(f"File not found: {filepath}")
        return None
```

```python
def load_raw_files(raw_dir: Path) -> Dict[str, pd.DataFrame]:
    """
    Load all raw CSV files from the specified directory
    
    Args:
        raw_dir: Path to the raw data directory
    
    Returns:
        Dictionary with file names as keys and dataframes as values
    """
    logger.info(f"Loading raw files from: {raw_dir}")
    
    raw_data = {}
    
    # Define expected raw files
    expected_files = {
        'sales': 'sales.csv',
        'orders': 'orders.csv',
        'inventory': 'inventory.csv',
        'warehouses': 'warehouses.csv'
    }
    
    for name, filename in expected_files.items():
        filepath = raw_dir / filename
        if filepath.exists():
            try:
                df = pd.read_csv(filepath)
                raw_data[name] = df
                logger.info(f"  - Loaded {name}: {len(df)} rows, {len(df.columns)} columns")
            except Exception as e:
                logger.error(f"  - Error loading {filename}: {str(e)}")
        else:
            logger.warning(f"  - File not found: {filepath}")
    
    if not raw_data:
        logger.error("No files were loaded! Check the raw directory path.")
    
    return raw_data
```
